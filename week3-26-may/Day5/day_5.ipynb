```python
class Animal:
    def speak(self):
        print("voice")
class cat:
    def speak(Animal):
        print("Meo meo")
class wolf:
    def speak(self):
        print("ao ao")

my_object = [Animal(),cat(),wolf()]
for obj in my_object:
    obj.speak()
```

## over write

```python
from abc import ABC, abstractmethod
```

```python
class Animal:
    @abstractmethod
    def speak(ABC):
        print("voice")
class cat:
    def speak(Animal):
        print("Meo meo")
class wolf:
    def speak(self):
        print("ao ao")
def voice(animal: Animal):
    animal.speak()

c1 = cat()
w1 = wolf()
voice(w1)
```

```python
class Animal():
    def __init__(self,color,teeth):
        self.__color=color
        self.__teeth=teeth
    def speak(Animal):
        print(f"{self.color} with the {self.teeth})
class cat(Animal):
    def speak(self):
        print(f"{self.teeth} with the {self.color})

c1 = cat("white",24)
c1.speak()
```
