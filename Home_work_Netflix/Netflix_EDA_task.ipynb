### 2. Identify and handle missing values:  
o Use dropna() for critical fields.  
o Use fillna() where appropriate (e.g., fill missing country or rating).  
o Experiment with IterativeImputer for fields with numerical implications (like
release_year).  

```python
df.isnull().sum()
```

```python
# replace missing values with unknow
df['director'] = df['director'].fillna('Unkown')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
# drop country missing data as it is crtical for demographic analysis
df = df.dropna(subset=['country'])
```

```python
df['cast'] = df['cast'].fillna('Unknown')
```

```python
# drop with few missing values
df = df.dropna(subset=['date_added','rating'])
```

```python
df.isnull().sum()
```

### 3. Parse and clean columns:  
o Extract numeric values from duration.  
o Convert date_added to datetime format.  
o Create a new column for content age (current year – release year).  

```python
df.head()
```

```python
df.dtypes
```

```python
df['date_added'] = pd.to_datetime(df['date_added'], format='mixed')
```

```python
df.head()
```

```python
df['content_age'] = 2025 -df['release_year']
```

```python
df.head()
```

```python
df['Duration'] = df['duration'].str.extract(r'(\d+)\s*min', expand=False).fillna(0).astype(float)
```

```python
df['seasons'] = df['duration'].str.extract(r'(\d+)\s*Season',expand=False).fillna(0).astype(float)
```

```python
df.head()
```

```python
df['era'] = pd.cut(
    df['release_year'],
    bins=[1900, 1980, 2000, 2015, float('inf')],
    labels=['Classic Era', 'Late 20th Century', 'Early 21st Century', 'Modern Era']
)
```

```python
df.head()
```

```python
df['is new'] = df['release_year'] -2020 >= 0
```

```python
df['is new'] = df['is new'].astype(int)
```
