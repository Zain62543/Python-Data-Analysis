```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
np.random.seed(42)
X = 2*np.random.randn(100,1)
y = 3*X+4+np.random.randn(100,1)
```

```python
data = pd.DataFrame({'X':X.flatten(), 'y':y.flatten()})
```

```python
data.head(10)
```

```python
plt.scatter(data['X'],data['y'])
plt.xlabel('X')
```

```python
# Linear regression from scratch
x_mean = X.mean()
y_mean = y.mean()
```

```python
numerator = np.sum((X-x_mean)*(y-y_mean))
Denominator = np.sum((X-x_mean)*(X-x_mean))
```

```python
w = numerator/Denominator
print(w)
```

```python
b = y_mean - w*x_mean
print(b)
```

```python
y_pred_man = w*X+b
```

```python
plt.scatter(data['X'],data['y'])
plt.plot(X,y_pred_man,c='r')
```

```python
# sclearn 
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
lr.fit(X,y)
```

```python
y_pred = lr.predict(X)
```

```python
print(lr.coef_)
print(lr.intercept_)
```

```python
plt.scatter(X,y)
plt.plot(X,y_pred_man)
plt.plot(X,y_pred)
```

```python
mse = mean_squared_error(y,y_pred)
mse
```
