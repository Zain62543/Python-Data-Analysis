
# Netflix Movies and TV Shows: Exploratory Data Analysis

## Objective
Explore the Netflix catalogue to understand its content
composition, release trends, genres, countries, and ratings.

## Questions
1. How many movies and TV shows are in the dataset?
2. Which years have the most content releases?
3. Which countries contribute the most titles?
4. What are the most common genres?
5. How has Netflix content addition changed over time?
6. What data-quality issues need to be addressed?

## Tools
Python, Pandas, NumPy, Matplotlib, Seaborn, Jupyter Notebook


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)

DATA_PATH = Path("../data/netflix_titles.csv")

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully.")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())
```

```python

print("Column names:")
print(df.columns.tolist())

print("\nData types and non-null counts:")
df.info()

print("\nMissing values:")
missing = df.isnull().sum().sort_values(ascending=False)
display(missing[missing > 0].to_frame("Missing values"))

print("\nDuplicate rows:", df.duplicated().sum())

print("\nDuplicate show IDs:", df["show_id"].duplicated().sum())
```
