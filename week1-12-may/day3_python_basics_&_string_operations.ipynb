## String formating

```python
#using formate() method
name = input("What is your name: ")
place = input("what is place: ")
my_string = "helo {} how are you {}".format(name,place)
print(my_string)
```

```python
my_string = "helo {} how are you {}".format("zain","IT")
print(my_string)
```

```python
#using f-string
message = f"hello, {name} how is you {place}"
print(message)
```
