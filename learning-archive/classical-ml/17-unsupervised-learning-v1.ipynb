## 1-K_means nearest neigbour

```python
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Generate synthetic 2D data
X, y_true = make_blobs(n_samples=1000, centers=8, cluster_std=2.0,random_state=42)
```

```python
# plot
plt.figure(figsize=(8,6))
plt.scatter(X[:,0],X[:,1], s=100)
plt.title("synthetic data cluster")
plt.grid(True)
plt.show()
```

```python
# KMeans
from sklearn.cluster import KMeans
```

```python
# Fit kmeans
kmeans = KMeans(n_clusters=4,random_state=42)
kmeans.fit(X)
y_kmeans = kmeans.predict(X)
```

```python
# Fit Kmeans
plt.figure(figsize=(8,6))
plt.scatter(X[:,0],X[:,1], c = y_kmeans, cmap='viridis')
plt.scatter(kmeans.cluster_centers_[:,0],kmeans.cluster_centers_[:,1], s=100, c='red',marker='+')
plt.title("KMeans clustering on synthetic data")
plt.legend()
plt.grid(True)
plt.show()
```

```python
inertias = []
k_range = range(1,11)

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(X)
    inertias.append(km.inertia_)
```

```python
plt.figure(figsize=(8,6))
plt.plot(k_range,inertias,'bo',linestyle = '-')
plt.xlabel("number of clusters")
plt.grid(True)
```

```python
from sklearn.datasets import load_wine
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
# data
data = load_wine()
X = data.data
y = data.target
```

```python
# scaling

sc = StandardScaler()
X_scaled = sc.fit_transform(X)
```

```python
#PCA
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)
```

```python
plt.scatter(X_pca[:,0],X_pca[:,1])
```

```python
# Fit kmeans
kmeans = KMeans(n_clusters=3,random_state=42)
kmeans.fit(X_pca)
y_kmeans = kmeans.predict(X_pca)
```

```python
# Fit Kmeans
plt.figure(figsize=(8,6))
plt.scatter(X_pca[:,0],X_pca[:,1], c = y_kmeans, cmap='viridis')
plt.scatter(kmeans.cluster_centers_[:,0],kmeans.cluster_centers_[:,1], s=100, c='red',marker='+')
plt.title("KMeans clustering on synthetic data")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Elbow plotting
inertias = []
k_range = range(1,11)

for k in k_range:
    km = KMeans(n_clusters=k, n_init=10,random_state=42)
    km.fit(X_pca)
    inertias.append(km.inertia_)
```

```python
plt.figure(figsize=(8,6))
plt.plot(k_range,inertias,'bo',linestyle = '-')
plt.xlabel("number of clusters")
plt.grid(True)
```
