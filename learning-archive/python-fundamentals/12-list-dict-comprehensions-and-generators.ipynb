```python
my_list = [12,45,67,87,34]
```

```python
for i in range(len(my_list)):
    my_list[i] = 2*my_list[i]
```

```python
my_list2 = [1,2,3,4,5,6,7]
new_list = []
for i in my_list2:
    new_list.append(i+5)

print(new_list)
```

```python
my_list = []
for i in range(12,33,5):
    my_list.append(i-5)
print(my_list)
```

```python
my_list = [i for i in range(15)]
print(my_list)
```

```python
type(my_list)
```

```python
pairs_1 = []
for i in range(1,3):
    for j in range(1,3):
        pairs_1.append((i,j))

print(pairs_1)
```

```python
pairs_1 = [(i,j) for i in range(1,4) for j in range(2,3)]
print(pairs_1)
```

# conditions in comprehension

```python
my_list = [12,45,67,21,34,55,66]
my_list = [i if my_list[i]%10 <=5]
print(my_list)
```

```python
my_list = [12,45,67,21,34,55,66]
```

```python
my_list = [120,450,670,210,340,550,660]
my_list = [i for i in my_list if i%100 <50]
print(my_list)
```

# dictionary comrehension

```python
my_dict = {num: -num for num in range(0,11)}
print(my_dict)
```

```python
my_dict = {num: -num*5 for num in range(0,11)}
print(my_dict)
```

```python
type(my_dict.keys[])
```

```python
my_dict.keys(3)
```

```python
# dictionary of even and odds
my_dict = {i-1:i for i in range(1,20,2)}
print(my_dict)
```

```python
my_dict = {i-1:i for i in range(1,20,2)}
print(my_dict)
```

```python
my_list = (i for i in range(10))
print(my_list)
```

```python
for i in my_list:
    print(i)
```

 ### Generator

```python
def num_sequence(n):
    i = 0
    while i < n:
        yield i
        i += 1
```
