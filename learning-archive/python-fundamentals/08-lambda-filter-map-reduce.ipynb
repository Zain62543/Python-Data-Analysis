# Functional programing

```python
# use a mathematical function to perform same task as a loop will do (imperitive)
```

```python
# 1 pure funtion
# recursion
# Functions are first class 
# imutable
```

```python
# Pure
def sum(x,y):
    return x+y
```

```python
sum(2,3)
```

```python
# not pure
total = 0
def test(a):
    global total
    total = total +a
    return total
```

```python
test(2)
```

```python
test(2)
```

### properties of Functional programing

### Filter function : filter()

```python
# filter(function, iterable)
my_list = [12,45,67,43]
even = filter(lambda x:x%2==0,my_list)
print(list(even))
```

```python
# filter(function, iterable)
my_list = [12,45,67,43]
odd = filter(lambda x:x%2==1,my_list)
print(list(odd))
```

### MAP funtion

```python
# filter(function, iterable)
my_list = [12,45,67,43]
odd = map(lambda x:x**2,my_list)
print(list(odd))
```

```python
celsius_temperaure = [0,10,20,30]
# find fahrenheit temperatures
fahrenheit_temp = map(lambda x:x*(9/5)+32,celsius_temperaure)
print(list(fahrenheit_temp))
```

### reduce function

```python
from functools import reduce
```

```python
# add two parameters to make it single
numbers = [1,2,3,4,5]
sum_result = reduce(lambda x,y: x+y, numbers)
print("sum",sum_result)
```
