# List

```python
my_list = [12,45,67,80,24]
square = [x*x for x in my_list]
print(square)
```

```python
even = [x for x in my_list if x%2 == 0]
print(even)
```

```python
shoping = ["glass_on_sale","shirt_on_sale","jeans"]
sale_list = [x for x in shoping if "sale" in x]
print(sale_list)
```
