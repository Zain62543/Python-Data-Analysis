## Destructor

```python

# destructor function is used to release data and delete an object
class car:
    def __init__(self,color,model,engine):
        self.__color = color
        self.__model = model
        self.__engine = engine
        self.number_of_tyres = 4
    def __del__(self):
        print("object destroyed")
    def start(self):
        print(f"{self.__color} car having {self.__model} with {self.__engine} has {self.number_of_tyres}")
toyota = car("red","2023","Electric")
toyota.start()
del toyota
```

```python
class vehicle:
    def __init__(self,make,model,year):
        self.make = make
        self.model = model
        self.year = year
    def vehicle_info(self):
        print(f"this car is {self.make}, {self.model} and {self.model}")
toyota = vehicle("x","y","z")
toyota.vehicle_info()

        
```
