```python
#repeating actions
#do, while, for in c++
# for , while in python
# while for undefined termination, not sure about end, runs as long condition is true.
c = 0
while c < 5:
    print(c)
    c = c+1
print("end")

# for for known termination
my_list = ["Apple","mangoes","kiwi","tomato"]
for i in my_list:
    print(i)
#range function
for i in range(1,6):
    print(i)

for i in range(10,1,-1):
    print(i)
```

## Nested loop

```python
matrix =[[1,2,3],[4,5,6],[67,87,91]]
print(matrix)
for rows in matrix:
    for column in rows:
        print(column,end=" ")
    else:
        print("\n")
# loop 2
for i in range(1,10):
    for j in range(1,6):
        print(f"{i*j:4}",end='')
    print()
```

## control loop

```python
#control the flow of loops
#modify the default behaviour of loops 
# types of control statements:
#Break exits the loops immmediatly
#continue : skip the rest of cuurent iteration and moves to the next iteration
#Pass: Does nothing; acts as placeholder
for i in range(10):
    if i>5:
        break
    print(i)

#2
for i in range(10):
    if i == 5:
        continue
    print(i)
#3
for i in range(10):
    if i < 5:
        pass
    print(i)

```
