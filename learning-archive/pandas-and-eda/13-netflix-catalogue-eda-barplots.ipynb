
# Netflix Movies and TV Shows: Exploratory Data Analysis

## Objective
Explore the Netflix catalogue to understand its content
composition, release trends, genres, countries, and ratings.

## Questions
1. How many movies and TV shows are in the dataset?
2. Which years have the most content releases?
3. Which countries contribute the most titles?
4. What are the most common genres?
5. How has Netflix content addition changed over time?
6. What data-quality issues need to be addressed?

## Tools
Python, Pandas, NumPy, Matplotlib, Seaborn, Jupyter Notebook


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)

DATA_PATH = Path("../../data/netflix_titles.csv")

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully.")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())
```

```python

print("Column names:")
print(df.columns.tolist())

print("\nData types and non-null counts:")
df.info()

print("\nMissing values:")
missing = df.isnull().sum().sort_values(ascending=False)
display(missing[missing > 0].to_frame("Missing values"))

print("\nDuplicate rows:", df.duplicated().sum())

print("\nDuplicate show IDs:", df["show_id"].duplicated().sum())
```

```python

netflix = df.copy()

# Remove accidental whitespace from text fields
text_cols = netflix.select_dtypes(include="object").columns
for col in text_cols:
    netflix[col] = netflix[col].str.strip()

# Remove duplicate rows
netflix = netflix.drop_duplicates()

# Convert date_added into a datetime column
netflix["date_added"] = pd.to_datetime(
    netflix["date_added"],
    errors="coerce"
)

# Convert release_year to numeric where possible
netflix["release_year"] = pd.to_numeric(
    netflix["release_year"],
    errors="coerce"
)

print("Original rows:", len(df))
print("Rows after removing duplicate records:", len(netflix))
print("\nRemaining missing values:")
display(
    netflix.isnull().sum()
    .sort_values(ascending=False)
    .to_frame("Missing values")
)
```

```python

type_counts = netflix["type"].value_counts()

ax = sns.barplot(
    x=type_counts.index,
    y=type_counts.values
)

ax.set_title("Netflix Catalogue: Movies vs TV Shows")
ax.set_xlabel("Content type")
ax.set_ylabel("Number of titles")

for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()
plt.show()
```
