# Text Generation with GPT-2 (HuggingFace Transformers)

In this example, we'll load a pre-trained GPT-2 model and use it to generate text based on an input prompt.

```python
# Import required classes
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch
```

```python
# Load the GPT-2 tokenizer and model
tokenizer = AutoTokenizer.from_pretrained("gpt2")
model = AutoModelForCausalLM.from_pretrained("gpt2")
```

```python
tokenizer = AutoTokenizer.from_pretrained("EleutherAI/gpt-j-6B")
model = AutoModelForCausalLM.from_pretrained("EleutherAI/gpt-j-6B")
```

```python
# Set your prompt/query
prompt = "GIKI F7 IT Centre is "
```

```python
# Tokenize the input and convert to tensor
inputs = tokenizer(prompt, return_tensors="pt")
```

```python
# Generate continuation of the text
outputs = model.generate(
    **inputs,
    max_length=100,         # Max total length (prompt + output)
    num_return_sequences=1, # Number of responses
    do_sample=True,         # Enable sampling (for randomness)
    top_k=50,               # Top-k sampling
    top_p=0.95,             # Nucleus sampling
    temperature=0.9         # Controls randomness (higher = more random)
)
```

```python
# Decode and print the generated text, GPT-2
generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
print(generated_text)
```

```python
from transformers import ViTFeatureExtractor, ViTForImageClassification
from PIL import Image
import requests
import torch
import matplotlib.pyplot as plt

# Load age classification model
model_name = "nateraw/vit-age-classifier"
model = ViTForImageClassification.from_pretrained(model_name)
feature_extractor = ViTFeatureExtractor.from_pretrained(model_name)

# Put model in eval mode
model.eval()
```

```python
# Use image from URL
#image_url = "https://images.unsplash.com/photo-1544005313-94ddf0286df2"  # Sample human face
image_url = "https://t4.ftcdn.net/jpg/02/45/56/35/360_F_245563558_XH9Pe5LJI2kr7VQuzQKAjAbz9PAyejG1.jpg"
image = Image.open(requests.get(image_url, stream=True).raw).convert("RGB")

# OR load local image
#image = Image.open("/content/HSAQ_PP_2023_AUG.png").convert("RGB")

# Display
plt.imshow(image)
plt.axis("off")
plt.title("Input Image")
plt.show()
```

```python
# Preprocess
inputs = feature_extractor(images=image, return_tensors="pt")

# Forward pass
with torch.no_grad():
    outputs = model(**inputs)

# Get predicted class
logits = outputs.logits
predicted_class_idx = logits.argmax(-1).item()
predicted_label = model.config.id2label[predicted_class_idx]

print("Predicted Age Group:", predicted_label)
```
