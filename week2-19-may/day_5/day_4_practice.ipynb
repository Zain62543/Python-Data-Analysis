```python
n = int(input("give a number: "))
mul = 1
for i in range(1,n+1):
    mul = mul*i
print(mul)
```

```python
#factorial
def recursion(n):
    if n == 0:
        return 1
    else:
        return n*recursion(n-1)
    
```

```python
recursion(6)
```

```python
#area of circle
import math
r = int(input("radius r"))
area = math.pi*(r**2)
print(round(area,2))
```

```python
# area by funtion
import math
def Area_of_circle():
    return math.pi*(r**2)
r = int(input("radius r"))
round(Area_of_circle(),3)
```

# parameters and arguments

```python
#x,y parameters , 2,3 arguments
# position of argument matters. corresponding arguments will be take place according to parametr's postion
```

```python
def my_func(greet,name):
    print(f"{greet},{name}")
my_func("Asma","Hi")
```

```python
def my_func(a,b):
    print(f"{b}")
    return a
my_func(b=1,a="Asma")
```

```python
#variable arguments
def add(*n):
    return list(map(lambda x:x**2,n))
add(1,2,8)
```

```python
details = {"name":"manan", "age" : 27, "qual":"eng"}
# taking dictionay as arguments
def my_fun(**details):
    print(details)
```
